*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here
#import numpy as np

#check function
def check_game(matrix):
    n = len(matrix)

    #check for rows
    for i in range(n):
        if matrix[i][0] != 0 and all(matrix[i][j] == matrix[i][0] for j in range(n)):
            return matrix[i][0]

    # matrix [, O, X]
    #        [, O, ]
    #        [, O, X]

    #check for columns
    for j in range(n):
        if matrix[0][j] != 0 and all(matrix[i][j] == matrix[0][j] for i in range(n)):
            return matrix[0][j]

    #check first diagonal
    if matrix[0][0] != 0 and all(matrix[i][i] == matrix[0][0] for i in range(n)):
        return matrix[0][0]

    #check reserce diagonal
    if matrix[0][n-1] != 0 and all(matrix[i][n-1-i] == matrix[0][n-1] for i in range(n)):
        return matrix[0][n-1]

    #check for empty spaces
    for row in matrix:
        if 0 in row:
            return -1

    #no winner and no empty spaces = draw
    return 0

#getting n
#n= int(input("Enter a number: "))

#building the rows and columns 
matrix = [[1, 1, 1],
          [2, 0, 2],
          [0, 0, 0]]
#for i in range(n):
#    row = []
#    for j in range(n):
#        row.append(" ")
#    matrix.append(row)
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
# Write your solution here
def place_mark(matrix, player_number, row, col):
    if player_number == 1:
        mark = "X"
    else:
        mark = "O"

    if(matrix[row][col] == 0):
        matrix[row][col] = mark
    else:
        return False

    return True
    
````

````python
# Write your solution here
def print_board(matrix):
    n = len(matrix)

    #column labels
    print("  ", end="")
    for i in range(n):
        print(i + 1, end="  ")
    print()

    #rows
    for i in range(n):
        row_letter = chr(65 + i)
        print(row_letter, end=" ")

        for j in range(n):
            print(matrix[i][j], end="  ")

        print()
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.


````python
# Write your solution here
def player_move(matrix, player_number):
    n = len(matrix)

    while True:
        location = input("Player " + str(player_number) + ", enter a location: ").upper()

        #check location with coord
        if len(location) < 2:
            print("Invalid location. Try again.")
            continue

        row_letter = location[0]

        try:
            col_number = int(location[1:])
        except:
            print("Invalid location. Try again.")
            continue

        row = ord(row_letter) - 65

        #check location on board
        if row < 0 or row >= n or col_number < 1 or col_number > n:
            print("Invalid location. Try again.")
            continue

        if make_move(matrix, player_number, location):
            return True
        else:
            print("That spot is already taken. Try again.")
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write your solution here
def minimax(matrix, depth, max_depth, maximizing):
    status = check_game(matrix)

    if status == "O":
        return 10 - depth

    if status == "X":
        return depth - 10

    if status == 0:
        return 0

    if depth >= max_depth:
        return 0

    n = len(matrix)

    if maximizing:
        best_score = -1000

        for i in range(n):
            for j in range(n):
                if matrix[i][j] == 0:
                    matrix[i][j] = "O"

                    score = minimax(matrix, depth + 1, max_depth, False)

                    matrix[i][j] = 0

                    if score > best_score:
                        best_score = score

        return best_score

    else:
        best_score = 1000
    
        for i in range(n):
            for j in range(n): 
                if matrix[i][j] == 0:
                    matrix[i][j] = "X"
    
                    score = minimax(matrix, depth + 1, max_depth, True)
    
                    matrix[i][j] = 0

                    if score < best_score:
                        best_score = score
                    
        return best_score

def smart_computer_move(matrix, max_depth):
    n = len(matrix)

    best_score = -1000
    best_row = -1
    best_col = -1

    for i in range(n):
        for j in range(n):
            if matrix[i][j] == 0:
                matrix[i][j] = "O"

                score = minimax( matrix, 0, max_depth, False)

                matrix[i][j] = 0

                if score > best_score:
                    best_score = score
                    best_row = i
                    best_col = j

        if best_row != -1:
            matrix[best_row][best_col] = "O"
            return True
            
        return False
````
